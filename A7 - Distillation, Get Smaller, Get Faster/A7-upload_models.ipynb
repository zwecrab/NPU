```python
from transformers import AutoModelForSequenceClassification, AutoTokenizer
from huggingface_hub import HfApi

# Initialize HF API
api = HfApi()

# For each model
for model_type in ["Odd-Layer", "Even-Layer"]:
    # Local path to your model
    local_path = f"./{model_type}_best"
    
    # HF repository name (change to your username)
    repo_name = f"st125338/{model_type.lower()}-toxic-comments"
    
    # Create repository
    api.create_repo(
        repo_id=repo_name,
        repo_type="model",
        exist_ok=True
    )
    
    # Upload files
    api.upload_folder(
        folder_path=local_path,
        repo_id=repo_name,
        repo_type="model"
    )
    
    # Upload tokenizer (if saved with model)
    try:
        tokenizer = AutoTokenizer.from_pretrained(local_path)
        tokenizer.push_to_hub(repo_name)
    except:
        print(f"Tokenizer not found for {model_type}")
```
